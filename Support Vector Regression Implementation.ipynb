```python
## featue encoding(Label Encoding and One Hot Encoding)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
## indepenedent and Deprndent feature
```

```python
df.columns
```

```python
x=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y=df['total_bill']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
x_train.head()
```

```python

x_train['sex']=le1.fit_transform(x_train['sex'])
x_train['smoker']=le2.fit_transform(x_train['smoker'])
x_train['time']=le3.fit_transform(x_train['time'])
```

```python
x_train.head()
```

```python
x_test['sex'] = le1.transform(x_test['sex'])       # Transforms 'sex' column
x_test['smoker'] = le2.transform(x_test['smoker']) # Transforms 'smoker' column
x_test['time'] = le3.transform(x_test['time'])     # Transforms 'time' column
```

```python
x_test.columns
```

```python
## One Hot Encoding -- column Transformer

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[3])],remainder='passthrough')
```

```python
x_train=ct.fit_transform(x_train)
```

```python
x_test=ct.transform(x_test)
```

```python
x_test
```

```python
### Apply support vector regressor
###SVR

from sklearn.svm import SVR
svr=SVR()
```

```python
y_pred=svr.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
```
